```python
import pandas as pd
df_clientes = pd.read_csv('../data/hey_clientes_clean.csv')
df_productos = pd.read_csv('../data/hey_productos_clean.csv')
df_transacciones = pd.read_csv('../data/hey_transacciones_clean.csv')


df_merged = pd.merge(
    df_transacciones,
    df_productos,
    on='user_id',
    how='left',
    suffixes=('_trans', '_prod')
)

# 2. Unir el resultado con clientes
#    user_id vincula todo a nivel cliente
df_unificada = pd.merge(
    df_merged,
    df_clientes,
    on='user_id',
    how='left',
    suffixes=('', '_cliente')
)

# 3. Resultado: una fila por transacción con todos los campos
print(f"Transacciones finales: {len(df_unificada)}")
print(f"Columnas totales: {len(df_unificada.columns)}")
df_unificada.head()
```

```python
# Mostrar todas las columnas (sin límite)
pd.set_option('display.max_columns', None)

# Mostrar todas las filas (sin límite)
pd.set_option('display.max_rows', None)

# Para que el ancho de cada columna no se corte
pd.set_option('display.max_colwidth', None)

# Para que no use notación científica en los números
pd.set_option('display.float_format', lambda x: '%.3f' % x)

# Ahora sí, ver todos los dtypes sin cortes
df_unificada.dtypes
```

```python
df_unificada.to_csv('../data/hey_data_unificada.csv', index=False)
```
